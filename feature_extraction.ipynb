## Bag of words

Simple and intutive
has fixed size 
Captures little bit semantic meaning

##### Disadvantage 
Sparsity -> vector contains lot of zer values 
overfitting 
fils to handle out of vocabulary word

## Bag of n-grams
it will try to capture the meaning of the word by considering n tokens as one word 

example 
D1: people watch campusx
D2: campusx watch campusx
D3: people write comment 
D4: campusx write comment 


if we consider bag of bigrams, then vector will be
    people watch | watch campusx| campux watch | people write| write comment | campusx write 
D1  1|1|0|0|0|0 
D2  0|1|1|0|0|0 
D3  0|0|0|1|1|0 
D4  0|0|0|0|1|1 

Advantage 
Able to capture semantics
easy to implement 

Disadvantage 
dimension of vocabular increases and time complexity increases 
not handles out of vocabulary words

## TF-IDF (Term Frequency - Inverse document frequency)
Any word occures mor frequently in particular document but other document frequency is very less 
TF(t,d) = (Number of occurences of term t in document d)
        -------------------------------------------------   (Total number of terms in document d)

IDF(t) = log (Total number of documnets in the corpus)
             -----------------------------------------
             (Number of documents with term t in them)

TF ranges from 0-1
Tf-IDF = TF*IDF



=========================
Reason to take log in IDF : to normalize the value 

Example if we have 10000 documents and we have very rare word that only present in 1 document then , IDF without log = 10000, where as with log it will be 4 , if we multiply 10000 with tf , the iDF dominates TF , so to avoid it we are normalizing IDF using log

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer()
tfidf.fit_transform(df['text']).toarray()
```

```python
print(tfidf.idf_)
```

```python
print(tfidf.get_feature_names_out())
```
